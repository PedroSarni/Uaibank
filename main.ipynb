```python
import pandas as pd
import numpy as np
import unicodedata 
```

```python
df=pd.read_csv("/home/sarni/Área de trabalho/1º Semestre/UaiBank/Projeto/data/Usuarios.csv")
df=df.set_index(["ID"])
df_transf=pd.read_csv("/home/sarni/Área de trabalho/1º Semestre/UaiBank/Projeto/data/Transferências.csv")

#tratamento
def removendo_acentos(texto_acentos):
    sem_acento=unicodedata.normalize('NFKD', texto_acentos)
    return ''.join([x for x in sem_acento if not unicodedata.combining(x)])

df["Nome"]=df["Nome"].str.strip()
df["Nome"]=df["Nome"].str.title()
df["Estado"]=df["Estado"].str.upper()
df["Estado"]=df["Estado"].apply(removendo_acentos)

estados_brasileiros = {
    "ACRE": "AC",
    "ALAGOAS": "AL",
    "AMAPA": "AP",
    "AMAZONAS": "AM",
    "BAHIA": "BA",
    "CEARA": "CE",
    "DISTRITO FEDERAL": "DF",
    "ESPIRITO SANTO": "ES",
    "GOIAS": "GO",
    "MARANHAO": "MA",
    "MATO GROSSO": "MT",
    "MATO GROSSO DO SUL": "MS",
    "MINAS GERAIS": "MG",
    "PARA": "PA",
    "PARAIBA": "PB",
    "PARANA": "PR",
    "PERNAMBUCO": "PE",
    "PIAUI": "PI",
    "RIO DE JANEIRO": "RJ",
    "RIO GRANDE DO NORTE": "RN",
    "RIO GRANDE DO SUL": "RS",
    "RONDONIA": "RO",
    "RORAIMA": "RR",
    "SANTA CATARINA": "SC",
    "SAO PAULO": "SP",
    "SERGIPE": "SE",
    "TOCANTINS": "TO"
}
regioes_brasileiras = {
    "AC": "NORTE",
    "AP": "NORTE",
    "AM": "NORTE",
    "PA": "NORTE",
    "RO": "NORTE",
    "RR": "NORTE",
    "TO": "NORTE",

    "AL": "NORDESTE",
    "BA": "NORDESTE",
    "CE": "NORDESTE",
    "MA": "NORDESTE",
    "PB": "NORDESTE",
    "PE": "NORDESTE",
    "PI": "NORDESTE",
    "RN": "NORDESTE",
    "SE": "NORDESTE",

    "DF": "CENTRO-OESTE",
    "GO": "CENTRO-OESTE",
    "MT": "CENTRO-OESTE",
    "MS": "CENTRO-OESTE",

    "ES": "SUDESTE",
    "MG": "SUDESTE",
    "RJ": "SUDESTE",
    "SP": "SUDESTE",

    "PR": "SUL",
    "RS": "SUL",
    "SC": "SUL"
}
siglas = set(estados_brasileiros.values())

df["Estado"] = df["Estado"].where(df["Estado"].isin(siglas),df["Estado"].map(estados_brasileiros))
df["Região"] = df["Estado"].map(regioes_brasileiras)
display(df_transf.style.hide(axis="index"))



display(df)
```
